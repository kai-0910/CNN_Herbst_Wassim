```python
import keras
from keras import layers
import tensorflow as tf
import tensorflow_datasets as tfds
import os
import numpy as np
import random

# We set the seed to be able to reproduce the results.
seed=1234
os.environ['PYTHONHASHSEED']=str(seed)
tf.random.set_seed(seed)
np.random.seed(seed)
random.seed(seed)
```

```python
# Loading the data set with a 80/20 split for training and validation, and the test set as is.
ds_train = tfds.load('stl10', split='train[:80%]',  as_supervised=True).batch(128)
ds_val = tfds.load('stl10', split='train[80%:]', as_supervised=True).batch(128)
ds_test = tfds.load('stl10', split='test', as_supervised=True).batch(128)
```

```python
# Checking the shape and data type of the images in the training set.
for images, labels in ds_train.take(1):
    print("Shape of images:", images.shape)
    print("datatyp of images:", images.dtype)
```

Preprocessing the data

```python
# change data typ to float
ds_train = ds_train.map(lambda img, lbl: (tf.cast(img, tf.float32), lbl))
ds_val   = ds_val.map(lambda img, lbl: (tf.cast(img, tf.float32), lbl))
ds_test  = ds_test.map(lambda img, lbl: (tf.cast(img, tf.float32), lbl))

# Check data type after change
for images, labels in ds_train.take(1):
    print("Shape of images:", images.shape)
    print("datatyp of images:", images.dtype)
```

```python
# Normalize the data
ds_train = ds_train.map(lambda img, lbl: (img / 255.0, lbl))
ds_val   = ds_val.map(lambda img, lbl: (img / 255.0, lbl))
ds_test  = ds_test.map(lambda img, lbl: (img / 255.0, lbl))

# Check data aber normalization
for img_batch, lbl_batch in ds_train.take(1):
    print("Minimum value:", tf.reduce_min(img_batch).numpy())
    print("maximum value:", tf.reduce_max(img_batch).numpy())
```

```python
# Categorize the labels
ds_train = ds_train.map(lambda img, lbl: (img, tf.one_hot(lbl, depth=10)))
ds_val   = ds_val.map(lambda img, lbl: (img, tf.one_hot(lbl, depth=10)))
ds_test  = ds_test.map(lambda img, lbl: (img, tf.one_hot(lbl, depth=10)))

# Check the categorize of the labels
for img_batch, lbl_batch in ds_train.take(1):
    print("Label shape:", lbl_batch.shape) 
    print("Example label:", lbl_batch[0].numpy()) 
```
