```python
import os
import random
import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
from keras import layers, models

# seed for reproducibility
seed = 1234
os.environ['PYTHONHASHSEED'] = str(seed)
tf.random.set_seed(seed)
np.random.seed(seed)
random.seed(seed)
```

```python
# load data
ds_train = tfds.load('stl10', split='train[:80%]', as_supervised=True).batch(128)
ds_val   = tfds.load('stl10', split='train[80%:]', as_supervised=True).batch(128)
ds_test  = tfds.load('stl10', split='test', as_supervised=True).batch(128)

# Query before preprocessing
print("Before preprocessing:")
for img_batch, lbl_batch in ds_train.take(1):
    print("Shape of images:", img_batch.shape)
    print("Data type of pictures:", img_batch.dtype)
    print("Max pixel value:", tf.reduce_max(img_batch).numpy())
    print("Example label (number):", lbl_batch[0].numpy())
```

```python
# function for preprocessing steps
def preprocess(img, lbl):
    img = tf.image.resize(img, [96, 96]) # Standardizing size (and implicit cast to float32)
    img = tf.cast(img, tf.float32)       # Change data type to float
    img = img / 255.0                    # Normalizing values [0, 1]
    lbl = tf.one_hot(lbl, depth=10)      # Transforming labels to categorical
    return img, lbl

# Apply the function to our data sets
ds_train = ds_train.map(preprocess)
ds_val   = ds_val.map(preprocess)
ds_test  = ds_test.map(preprocess)

# Query after preprocessing
print("After preprocessing")
for img_batch, lbl_batch in ds_train.take(1):
    print("Shape of images:", img_batch.shape)
    print("Data type of images:", img_batch.dtype)
    print("Max pixel value:", tf.reduce_max(img_batch).numpy())
    print("Shape of labels (Categorical):", lbl_batch.shape)
```
